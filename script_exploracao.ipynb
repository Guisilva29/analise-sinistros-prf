```python
import os

pasta_dados = '1_dados_brutos'
lista_dfs = []

# Mapeia e lê cada arquivo CSV da pasta
for arquivo in os.listdir(pasta_dados):
    if arquivo.endswith('.csv'):
        caminho_completo = os.path.join(pasta_dados, arquivo)
        df_temp = pd.read_csv(caminho_completo, sep=';', encoding='latin1')
        lista_dfs.append(df_temp)

# Concatena todas as tabelas em um único DataFrame
df_prf = pd.concat(lista_dfs, ignore_index=True)

df_prf
```

```python
df_prf.info()
```

```python

pasta_dados = '1_dados_brutos'

# Dicionário para guardar as colunas de cada arquivo
colunas_por_arquivo = {}

for arquivo in sorted(os.listdir(pasta_dados)):
    if arquivo.endswith('.csv'):
        caminho = os.path.join(pasta_dados, arquivo)
        # Lê apenas 1 linha para carregar só o cabeçalho rápido
        df_temp = pd.read_csv(caminho, sep=';', encoding='latin1', nrows=1)
        colunas_por_arquivo[arquivo] = list(df_temp.columns)

# Compara a quantidade de colunas em cada arquivo
for arquivo, cols in colunas_por_arquivo.items():
    print(f'{arquivo}: {len(cols)} colunas')
```

```python
# 1. Conta o total de nulos e calcula a porcentagem por coluna
nulos_qtd = df_prf.isna().sum()
nulos_pct = (df_prf.isna().mean() * 100).round(2)

# 2. Consolida em uma tabela, filtra apenas quem tem nulos e ordena do maior para o menor
relatorio_nulos = (
    pd.DataFrame({'Qtd_Nulos': nulos_qtd, 'Porcentagem_%': nulos_pct})
    .query('Qtd_Nulos > 0')
    .sort_values(by='Qtd_Nulos', ascending=False)
)

relatorio_nulos
```

```python
# Trata os valores nulos preenchendo com 'Não Informado'
colunas_nulas = ['uop', 'delegacia', 'regional', 'classificacao_acidente']
df_prf[colunas_nulas] = df_prf[colunas_nulas].fillna('Não Informado')

# Checa se existem linhas 100% duplicadas na base unificada
total_duplicadas = df_prf.duplicated().sum()
print(f"Total de linhas duplicadas: {total_duplicadas}")
```

```python
import sqlite3


# 1. Abre uma nova conexão com o arquivo do banco
conexao = sqlite3.connect('2_banco_de_dados/prf_database.db')

try:
    # 2. Envia a base bruta do Pandas para o SQLite
    df_prf.to_sql('tb_acidentes_bruto', conexao, if_exists='replace', index=False)
    
    # 3. Prepara e executa a criação da tabela filtrada
    cursor = conexao.cursor()
    cursor.execute("DROP TABLE IF EXISTS tb_acidentes_limpo;")
    
    query_criar_tabela = """
    CREATE TABLE tb_acidentes_limpo AS
    SELECT 
        id,
        data_inversa,
        dia_semana,
        horario,
        uf,
        br,
        km,
        municipio,
        causa_acidente,
        tipo_acidente,
        classificacao_acidente,
        fase_dia,
        sentido_via,
        condicao_metereologica,
        tipo_pista,
        tracado_via,
        uso_solo,
        pessoas,
        mortos,
        feridos_graves,
        feridos_leves,
        ilesos,
        feridos,
        veiculos,
        latitude,
        longitude
    FROM tb_acidentes_bruto;
    """
    cursor.execute(query_criar_tabela)
    conexao.commit()
    
    # 4. Lê as primeiras linhas para confirmar que a tabela foi criada
    df_teste = pd.read_sql_query("SELECT * FROM tb_acidentes_limpo LIMIT 5;", conexao)
    print("Tabela 'tb_acidentes_limpo' criada e validada com sucesso!")

finally:
    # 5. Garante o fechamento da conexão só no fim da execução
    conexao.close()

df_teste
```

```python
conexao = sqlite3.connect('2_banco_de_dados/prf_database.db')

# Extrair o ano da coluna data_inversa e contar sinistros por ano
# A coluna data_inversa possui datas em formatos diferentes, como "2023-01-15" e "15/01/2023"
# por isso, usamos uma expressão CASE para extrair o ano corretamente.

sinistros_por_ano = pd.read_sql_query('''
    SELECT 
        CASE 
            WHEN data_inversa LIKE '%-%' THEN SUBSTR(data_inversa, 1, 4)
            WHEN data_inversa LIKE '%/%' THEN SUBSTR(data_inversa, 7, 4)
            ELSE NULL
        END as ano,
        COUNT(id) as total_sinistros
    FROM tb_acidentes
    WHERE data_inversa IS NOT NULL
    GROUP BY ano
    ORDER BY total_sinistros DESC;
''', conexao)

sinistros_por_ano
```

```python
conexao = sqlite3.connect('2_banco_de_dados/prf_database.db')

# Para analisar a quantidade de vítimas fatais por tipo de acidente,
# podemos utilizar a função SUM para somar o número de mortos em cada tipo de acidente. 
# A consulta SQL abaixo agrupa os registros pelo tipo de acidente
# e calcula a soma das vítimas fatais, ordenando o resultado em ordem decrescente.
# Podemos observar que o tipo de acidente com maior número de vítimas fatais é a colisão,
#  seguida por saída de pista, atropelamento e capotamento.


vitimas_fatais_tipo_acidente = pd.read_sql_query('''SELECT tipo_acidente, sum(mortos) AS vitimas_fatais
 FROM tb_acidentes
 GROUP BY tipo_acidente ORDER BY vitimas_fatais DESC;''', conexao)

vitimas_fatais_tipo_acidente
```
